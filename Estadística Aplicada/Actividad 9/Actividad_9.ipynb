# **`Librerías`**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **`Lectura de Dataset`**

```python
link = 'https://raw.githubusercontent.com/Manumora117/AI-Master-s-Projects-and-Exercises/refs/heads/main/Estad%C3%ADstica%20Aplicada/Actividad%209/youtube_recommendation_dataset%20-.csv'
df = pd.read_csv(link)
display(df.info())
display(df.describe())
display(df.head(3))
```

# **`Preprocesamiento`**

```python
# @title Tratamiento de Nulos
suma = 0
for i in df.columns:
  nulos = df[i].isnull().sum()
  if nulos > 0:
    print(f'La columna {i} tiene {nulos} nulos')
    suma += 1
if suma == 0:
  print('¡No hay nulos!')
```

## **`Análisis Univariado`**

```python
# @title Numéricas
# Seleccion de columnas Numéricas
col_num = df.select_dtypes(include=["number"]).columns.tolist()

for col in col_num:
    print(f"\n===== Variable numérica: {col} =====")
    print(df[col].describe())
    print("Nulos:", df[col].isna().sum())

    # Histograma + densidad
    plt.figure(figsize=(8, 4))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Histograma y densidad de {col}")
    plt.xlabel(col)
    plt.ylabel("Frecuencia")
    plt.tight_layout()
    plt.show()

    # Boxplot para outliers
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot de {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# @title Categóricas
# Seleccion de columnas Categóricas
col_cat = df.select_dtypes(include=["object", "string", "category"]).columns.tolist()
if "Title" in col_cat:
    col_cat.remove("Title") # excluimos Title por ser muy alta cardinalidad

for col in col_cat:
    print(f"\n===== Variable categórica: {col} =====")
    vc = df[col].value_counts(dropna=False)
    print("Frecuencias absolutas:")
    print(vc)
    print("\nFrecuencias relativas (%):")
    print((vc / len(df) * 100).round(2))

    # Gráfico de barras (limitando a las 20 categorías más frecuentes para no saturar)
    top_n = 20
    plt.figure(figsize=(10, 4))
    order = vc.head(top_n).index
    sns.countplot(data=df, x=col, order=order)
    plt.title(f"Distribución de {col} (top {top_n})")
    plt.xlabel(col)
    plt.ylabel("Cuenta")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
# @title Limpieza y transformación más profunda
from scipy import stats
from sklearn.preprocessing import PowerTransformer

# 1. Selección de variables numéricas continuas
num_cols = [
    "view_count",
    "like_count",
    "comment_count",
    "engagement_rate",
    "likes_to_views_ratio",
    "comments_to_views_ratio",
    "duration_seconds",
    "video_age_days"
]

# 2. Tratamiento de outliers con z-score truncado
def winsorize_zscore(series, z_thresh=3.0):
    z = np.abs(stats.zscore(series.replace([np.inf, -np.inf], np.nan).dropna()))
    s = series.copy()
    # límites superior e inferior con base en percentiles "sanos"
    lower = series[(z < z_thresh)].quantile(0.01)
    upper = series[(z < z_thresh)].quantile(0.99)
    return np.clip(s, lower, upper)

for col in num_cols:
    df[col] = winsorize_zscore(df[col])

# 3. Corrección de sesgo (skewness) con transformación Yeo-Johnson
pt = PowerTransformer(method="yeo-johnson", standardize=False)
df[num_cols] = pt.fit_transform(df[num_cols])

df.head()
```

```python
# @title Extracción de variables categóricas y numéricas
cat_vars = df.select_dtypes(include=['object', 'category']).columns.tolist()
cat_num = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
```

```python
# @title Codificación
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Preprocesador: One-Hot para categóricas, estándar para numéricas
numeric_transformer = Pipeline(steps=[
    ("scaler", StandardScaler())])

categorical_transformer = Pipeline(steps=[
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, cat_num),
        ("cat", categorical_transformer, cat_vars)])
```

```python
# @title Estandarización
# Ajuste y transformación
X = df[cat_num + cat_vars].copy()
X_preprocessed = preprocessor.fit_transform(X)

# 4. Obtener nombres de columnas resultantes (útil para análisis posterior)
num_feature_names = cat_num
cat_feature_names = preprocessor.named_transformers_["cat"]["onehot"].get_feature_names_out(cat_vars)
feature_names = np.concatenate([num_feature_names, cat_feature_names])

df_model = pd.DataFrame(X_preprocessed, columns=feature_names, index=df.index)
df_model.head()
```

# **`Implementación del modelo: `**

```python
from sklearn.cross_decomposition import CCA
from sklearn.preprocessing import StandardScaler

# 1. Definir bloques de variables
# Bloque X: variables de "escala/alcance" del video
X_cols = [
    "view_count",
    "like_count",
    "comment_count",
    "duration_seconds",
    "video_age_days"]

# Bloque Y: variables de "eficiencia/engagement"
Y_cols = [
    "engagement_rate",
    "likes_to_views_ratio",
    "comments_to_views_ratio"]

# 2. Construir matrices X e Y
X = df_model[X_cols].values
Y = df_model[Y_cols].values

# 3. Ajustar CCA con k componentes canónicas
n_components = min(len(X_cols), len(Y_cols))
cca = CCA(n_components=n_components)

X_c, Y_c = cca.fit_transform(X, Y)

# 4. Correlaciones canónicas
corrs = []
for i in range(n_components):
    corr = np.corrcoef(X_c[:, i], Y_c[:, i])[0, 1]
    corrs.append(corr)

corrs = np.array(corrs)

# 5. Cargarings (correlaciones entre variables originales y variables canónicas)
loadings_X = np.corrcoef(X.T, X_c.T)[0:len(X_cols), len(X_cols):]
loadings_Y = np.corrcoef(Y.T, Y_c.T)[0:len(Y_cols), len(Y_cols):]

loadings_X_df = pd.DataFrame(
    loadings_X,
    index=X_cols,
    columns=[f"U{i+1}" for i in range(n_components)])

loadings_Y_df = pd.DataFrame(
    loadings_Y,
    index=Y_cols,
    columns=[f"V{i+1}" for i in range(n_components)])

# 6. DataFrame con correlaciones canónicas
cca_results = pd.DataFrame({
    "Componente": [f"Par {i+1}" for i in range(n_components)],
    "Correlacion_canonica": corrs})

print("Correlaciones canónicas:")
print(cca_results)

print("\nLoadings de X (variables originales vs U):")
print(loadings_X_df)

print("\nLoadings de Y (variables originales vs V):")
print(loadings_Y_df)
```
